```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/Student_Dataset.csv')

print('Dataset loaded successfully.')
```

### 1. Inspecting the Dataset

We will start by inspecting the first and last few rows of the DataFrame, as well as a few random samples, to get a preliminary sense of the data.

```python
print('First 5 rows of the dataset:')
display(df.head())

print('\nLast 5 rows of the dataset:')
display(df.tail())

print('\n5 random samples from the dataset:')
display(df.sample(5))
```

### 2. Dataset Shape and Information

Next, we will examine the dataset's dimensions, column names, index, and data types using `df.shape`, `df.columns`, `df.index`, and `df.info()`.

```python
print('Dataset shape (rows, columns):', df.shape)
print('Number of rows:', df.shape[0])
print('Number of columns:', df.shape[1])

print('\nColumn names:')
print(df.columns.tolist())

print('\nIndex:')
print(df.index)

print('\nDetailed information about the dataset (including non-null counts and dtypes):')
df.info()
```

### 3. Data Types Overview

Here's a direct look at the data types of each column.

```python
print('Data types of each column:')
display(df.dtypes)
```

### 4. Descriptive Statistics

Finally, we will generate descriptive statistics for numerical columns using `df.describe()` to understand the central tendency, dispersion, and shape of the dataset's distribution.

```python
print('Descriptive statistics for numerical columns:')
display(df.describe())
```
